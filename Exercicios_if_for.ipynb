```python
#Criando as listas iniciais
produtos = ["Camisa básica", "Calça Jeans","Jaqueta","Bermuda"]
precos = [49.90,129.90,219.90,59.90]
```

```python
#Cadastrando Novos itens
novo_produto = input("Informe o novo produto: ")
novo_preco = float(input("Informe o preço do novo produto: "))

produtos.append(novo_produto)
precos.append(novo_preco)

print(f"o produto {novo_produto} foi cadastrad com sucesso")

for produto, preco in zip(produtos, precos):
   print(f"O produto {produto}: custa R${preco:.2f}")
```

```python
# Remover o ultimo Item

if len(produtos)>0:
  produtos.pop()
  precos.pop()
  print(f"O produto {produto} com o preço {preco} foram removidos")

else:
  print("Estoque vazio")

for produto, preco in zip(produtos, precos):
   print(f"O produto {produto}: custa R${preco:.2f}")
```
